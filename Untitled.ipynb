```python
def create_mintpy_config_txt(work_dir, dem_path, inc_angle_path, az_angle_path, 
                        water_mask_path, lat_ref, lon_ref, output_path):
    """
    Create a flexible MintPy configuration file.
    
    Args:
        work_dir: Working directory path (e.g., "./stack")
        dem_path: Path to DEM file
        inc_angle_path: Path to incidence angle files
        az_angle_path: Path to azimuth angle files  
        water_mask_path: Path to water mask file
        lat_ref: Reference latitude (float)
        lon_ref: Reference longitude (float)
        output_path: Output path for the config file
    """
    
    CONFIG_TXT = f'''# vim: set filetype=cfg:
mintpy.load.processor      = aria  #[isce, aria, hyp3, gmtsar, snap, gamma, roipac], auto for isce
#---------interferogram datasets:
mintpy.load.unwFile        = {work_dir}/unwrapStack.vrt
mintpy.load.corFile        = {work_dir}/cohStack.vrt
mintpy.load.connCompFile   = {work_dir}/connCompStack.vrt
#---------geometry datasets:
mintpy.load.demFile        = {dem_path}
mintpy.load.incAngleFile   = {inc_angle_path}
mintpy.load.azAngleFile    = {az_angle_path}
mintpy.load.waterMaskFile  = {water_mask_path}
mintpy.reference.lalo           = {lat_ref}, {lon_ref}
mintpy.troposphericDelay.method = no
mintpy.deramp                   = no
mintpy.topographicResidual      = no
# options to speedup the processing (fast but not the best)
mintpy.networkInversion.weightFunc           = no
mintpy.topographicResidual.pixelwiseGeometry = no
'''
    
    write_config_file(output_path, CONFIG_TXT, mode='w')

# Example usage:
work_dir = f"{aria_files_path}/stack"
dem_path = f"{aria_files_path}/DEM/glo_90.dem"
inc_angle_path = f"{aria_files_path}/incidenceAngle/*.vrt"
az_angle_path = f"{aria_files_path}/azimuthAngle/*.vrt"
water_mask_path = f"{aria_files_path}/mask/esa_world_cover_2021.msk"
lat_ref = -23.25
lon_ref = -68.25

splitted_text = aria_files_path.split(os.path.sep)
if len(splitted_text)==1:
    print(aria_files_path.split('/'))
    splitted_text = aria_files_path.split('/')
folder_name = splitted_text[-1]

config_file = os.path.join(aria_files_path, f"mintpy_config_SdA_{folder_name}.txt")
create_mintpy_config_txt(work_dir, dem_path, inc_angle_path, az_angle_path,
                   water_mask_path, lat_ref, lon_ref, config_file)
```
